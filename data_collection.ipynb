```python
# Flatten the borough and district lists and create a mapping dictionary
district_to_borough = {}
for borough, districts in zip(borough_data['Borough'], borough_data['Districts']):
    for district in districts:
        district_to_borough[district] = borough

# Function to extract district from station name and map to borough
def get_borough(start_station_name):
    # Splitting on ', ' and taking the last element
    district = start_station_name.split(', ')[-1]
    # 'Unknown' if district not found
    return district_to_borough.get(district, 'Unknown')  

# Create new column 'borough'
filtered_df['borough'] = filtered_df['start_station_name'].apply(get_borough)

# Filter unknown rows
filtered_df = filtered_df.loc[~filtered_df['borough'].isin(['Unknown'])]
print(filtered_df.info())
```

# Normalisation

```python
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter out the outliers
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Apply the function to the 'revenue', 'roi', and 'budget' columns
filtered_df = remove_outliers(filtered_df, 'duration')

print(filtered_df.info())
```

# Aggregation station level

```python
grouped_df = filtered_df.groupby(['start_station_id', 'borough']).agg({'duration': ['count', 'mean']}).reset_index()
grouped_df.columns = ['station_id', 'borough', 'hires', 'avg_duration']
grouped_df['avg_duration'] = round((grouped_df['avg_duration']/60000),2)
```

# Data Quality Checks

```python
valid_boroughs = ['Westminster', 'Kensington and Chelsea', 'Camden', 'Islington', 'Greenwich', 
                  'Tower Hamlets', 'Hackney', 'Southwark', 'Lambeth', 'Wandsworth', 
                  'Hammersmith and Fulham', 'Newham']

# Perform data quality checks
def check_data_quality(df):
    errors = []
    
    # Check if 'hires' are positive and integers
    if not df['hires'].apply(lambda x: x > 0 and isinstance(x, int)).all():
        errors.append("Hires column contains non-positive values or non-integer values.")
    
    # Check if 'avg_duration' are positive and floats
    if not df['avg_duration'].apply(lambda x: x > 0 and isinstance(x, float)).all():
        errors.append("Avg_duration column contains non-positive values or non-float values.")

    # Check if 'station_id' is unique
    if not df['station_id'].is_unique:
        errors.append("Station_id column is not unique.")

    # Check if 'borough' is in inner London
    if not df['borough'].isin(valid_boroughs).all():
        errors.append("Borough column contains invalid entries.")

    return errors

# Perform the checks on grouped_df
data_quality_errors = check_data_quality(grouped_df)
data_quality_errors
```

```python
sex_df = pd.read_csv('/project/data_new/Data_SEX_AGE_UNIT.csv')

# Define the age groups
age_groups_corrected = {
    '0_4': '0_15', '5_9': '0_15', '10_14': '0_15', '15': '0_15',
    '16': '16_24', '17': '16_24', '18': '16_24', '19': '16_24', '20_24': '16_24',
    '25_29': '25_49', '30_34': '25_49', '35_39': '25_49', '40_44': '25_49', '45_49': '25_49',
    '50_54': '50_64', '55_59': '50_64', '60_64': '50_64',
    '65_and_over': '65_over'
}

labels = list(set(age_groups_corrected.values()))

# Initialise columns for each bin and gender for each borough in sex_df
for label in labels:
    sex_df[f"{label}_total"] = 0
    sex_df[f"{label}_males"] = 0
    sex_df[f"{label}_females"] = 0

# Perform the aggregation into the new bins, grouped by borough
for col in sex_df.columns:
    # Skip the borough column for aggregation
    if col == 'borough':
        continue
    
    # Split column name to extract age range and gender
    parts = col.split(' - ')
    if len(parts) == 3:
        age_part, sex_part, _ = parts
        age_key = '65_and_over' if '65 and over' in age_part.lower() else age_part.split(':')[1].strip().lower().replace('age ', '').replace(' to ', '_').replace('and over', 'over')
        bin_key = age_groups_corrected.get(age_key)

        sex_key = sex_part.split(':')[1].strip().lower().replace('total\\ sex', 'total')
        if bin_key and sex_key in ['total', 'males', 'females']:
            sex_df[f"{bin_key}_{sex_key}"] += sex_df[col]

# Drop old columns except for the borough column
columns_to_drop = [col for col in sex_df.columns if 'Age :' in col]
sex_df.drop(columns=columns_to_drop, inplace=True)

# Lowercasing column names
sex_df.columns = ['borough' if col == 'borough' else col.lower() for col in sex_df.columns]

sex_df.head()
```

# Add extra KPIs

```python
def summ(num1, *args):
    total = num1
    for num in args:
        total = total + num
    return total
```

```python
full_df['total_persons'] = summ(full_df['0_15_total'], full_df['16_24_total'], full_df['25_49_total'],
                                full_df['50_64_total'], full_df['65_over_total'])
full_df['hired_min_per_100k'] = round(full_df['hires'] * full_df['avg_duration']/(full_df['total_persons'])*100000,2)
full_df['male_percentage'] = round(100*summ(full_df['0_15_males'], full_df['16_24_males'], full_df['25_49_males'],
                                full_df['50_64_males'], full_df['65_over_males'])/full_df['total_persons'],2)
full_df['total_workforce'] = summ(full_df['occupation_1'], full_df['occupation_2'], full_df['occupation_3'],
                                full_df['occupation_4'], full_df['occupation_5'], full_df['occupation_6'], 
                                  full_df['occupation_7'], full_df['occupation_8'], full_df['occupation_9'])
full_df['bike_commuters_per_100k'] = round(full_df['no_bike_work_commute']/(full_df['total_persons'])*100000,2)
full_df['cars_per_100k'] = round(full_df['total_cars_vans']/(full_df['total_persons'])*100000,2)

station_counts = full_df.groupby('borough')['station_id'].nunique()
full_df['total_stations'] = full_df['borough'].map(station_counts)
full_df['total_stations_per_100k'] = round(full_df['total_stations']/(full_df['total_persons'])*100000,2)
full_df.head(12)
```
